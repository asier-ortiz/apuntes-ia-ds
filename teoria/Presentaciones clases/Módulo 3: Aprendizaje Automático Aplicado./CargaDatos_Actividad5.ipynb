```python
import pandas as pd
import os
from sklearn.utils import shuffle

# Simulación de carga del dataset como antes
base_dir = "/content/drive/MyDrive/UNIR/6/EuroSAT"
data = []

for class_name in os.listdir(base_dir):
    class_dir = os.path.join(base_dir, class_name)
    if os.path.isdir(class_dir):
        for img_file in os.listdir(class_dir):
            img_path = os.path.join(class_dir, img_file)
            data.append({"filepath": img_path, "label": class_name})

df = pd.DataFrame(data)
df = shuffle(df, random_state=42)
df
```

```python
def create_balanced_dataset(df, train_per_class, test_per_class):
    train_list = []
    test_list = []

    for label in df['label'].unique():
        class_subset = df[df['label'] == label]
        class_subset = shuffle(class_subset, random_state=42)

        # Asegurarse que hay suficientes imágenes por clase
        # total_needed = train_per_class + test_per_class
        # assert len(class_subset) >= total_needed, f"No hay suficientes imágenes para la clase {label}"

        train_samples = class_subset.iloc[:train_per_class]
        test_samples = class_subset.iloc[train_per_class:train_per_class + test_per_class]

        train_list.append(train_samples)
        test_list.append(test_samples)

    train_df = pd.concat(train_list).reset_index(drop=True)
    test_df = pd.concat(test_list).reset_index(drop=True)

    return train_df, test_df
```

```python
# Escenario 1: 100 por clase (train), 20 por clase (test)
train_df_1, test_df_1 = create_balanced_dataset(df, train_per_class=100, test_per_class=20)

# Escenario 2: 500 por clase (train), 100 por clase (test)
train_df_2, test_df_2 = create_balanced_dataset(df, train_per_class=500, test_per_class=100)
```

```python
import os
import shutil

def save_images_to_directory(train_df, test_df, base_output_dir):
    for split_name, split_df in [("train", train_df), ("test", test_df)]:
        for _, row in split_df.iterrows():
            label = row["label"]
            src_path = row["filepath"]
            dest_dir = os.path.join(base_output_dir, split_name, label)
            os.makedirs(dest_dir, exist_ok=True)

            dest_path = os.path.join(dest_dir, os.path.basename(src_path))
            shutil.copy2(src_path, dest_path)

    print(f"Imágenes copiadas en: {base_output_dir}")

save_images_to_directory(train_df_1, test_df_1, "escenario1")
save_images_to_directory(train_df_2, test_df_2, "escenario2")
```
