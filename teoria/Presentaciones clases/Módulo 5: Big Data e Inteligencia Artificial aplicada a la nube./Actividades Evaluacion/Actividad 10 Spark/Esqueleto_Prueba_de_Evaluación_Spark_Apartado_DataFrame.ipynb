## Descripción de Características incluidas en el archivo ene_mo20.csv

0. **PROVINCIA**: Tipo entero (código de la provincia).
1. **MUNICIPIO**: Tipo entero (código del municipio).
2. **ESTACION**: Tipo entero (número identificativo de la estación de medición).
3. **MAGNITUD**: Tipo entero (código que representa el tipo de magnitud medida, como contaminantes específicos).
4. **PUNTO_MUESTREO**: Tipo objeto (identificador compuesto del punto de muestreo).
5. **ANO**: Tipo entero (año de la medición).
6. **MES**: Tipo entero (mes de la medición).
7.  Hasta37. **D01-D31**: Tipo flotante (mediciones diarias del mes, una por cada día).
38. Hasta-68. **V01-V31**: Tipo objeto (validez de la medición diaria, 'V' para válido y 'N' para no válido).

### Cada par de columnas Dxx y Vxx corresponde a un día del mes y su respectiva validez.  Ejemplo:
- **D01**: Medición del día 1 del mes.
- **V01**: Validez de la medición del día 1 del mes.
...
- **D31**: Medición del día 31 del mes (si aplica).
- **V31**: Validez de la medición del día 31 del mes (si aplica).
