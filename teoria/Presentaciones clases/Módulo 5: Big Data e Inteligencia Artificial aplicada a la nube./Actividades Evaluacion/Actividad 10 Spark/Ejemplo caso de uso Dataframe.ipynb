```python
from pyspark.sql import SparkSession

# Crear o obtener una SparkSession existente
spark = SparkSession.builder \
    .appName("EjemplosDataFrame") \
    .getOrCreate()

# Verificar la SparkSession
print(spark)
```

```python
# spark is an existing SparkSession
df = spark.read.csv("churn.csv", header=True,inferSchema=True)
# Displays the content of the DataFrame to stdout
df.show(5)
```

```python
# spark, df are from the previous example # Print the schema in a tree format
df.printSchema()
```

```python
df.select(df['State'],df['Account_Length'],df['Area_Code'],df['Phone'],df['Intl_Plan'],df['Vmail_Plan'],df['Vmail_Message']).show()
```

```python
df.filter(df['State'] =='NY').show()
```

```python
df.groupBy("State").count().show()
```

```python
df.groupBy("Area_Code").count().show()
```

```python
from pyspark.sql.functions import col
from pyspark.sql.types import StringType,BooleanType,DateType
df.groupBy("State").sum("Day_Mins").filter(df['State']=='ID').show()
```
