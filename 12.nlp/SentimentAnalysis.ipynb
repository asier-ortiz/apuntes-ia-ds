```python
import tensorflow as tf
from tensorflow.keras import layers, models
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences

# --- Cargar y preparar el dataset IMDb ---
num_words = 10000  # Usamos las 10,000 palabras más frecuentes
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=num_words) # UNICAMENTE TIENEN INDICES DE FREQUENCIA

# --- Preprocesar los datos ---
max_length = 500  # Longitud máxima de las secuencias (en palabras)
x_train = pad_sequences(x_train, maxlen=max_length)
x_test = pad_sequences(x_test, maxlen=max_length)

# --- Construcción del modelo ---
model = models.Sequential()
model.add(layers.Embedding(num_words, 128, input_length=max_length))
model.add(layers.GlobalAveragePooling1D())
model.add(layers.Dense(64, activation='relu'))
model.add(layers.Dense(1, activation='sigmoid'))  # Salida binaria (positiva o negativa)

# --- Compilación del modelo ---
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

# --- Entrenamiento ---
model.fit(x_train, y_train, epochs=5, batch_size=128, validation_data=(x_test, y_test))

# --- Evaluación del modelo ---
loss, accuracy = model.evaluate(x_test, y_test)
print(f"Loss: {loss:.4f}, Accuracy: {accuracy:.4f}")

# --- Predicción de ejemplo ---
text_example = ["This movie was amazing! I loved it."]
# Convertir el texto de ejemplo a secuencia de números (como las secuencias de entrenamiento)
word_index = imdb.get_word_index()
encoded_example = [word_index.get(word, 0) for word in text_example[0].lower().split()]
encoded_example = pad_sequences([encoded_example], maxlen=max_length)

# Predecir el sentimiento del ejemplo
prediction = model.predict(encoded_example)
print(f"Sentiment prediction (0 = negative, 1 = positive): {prediction[0][0]:.4f}")
```
