# **Dataset: 20 Newsgroups**

El dataset **20 Newsgroups** es un conjunto de datos ampliamente utilizado en tareas de clasificación de texto y procesamiento de lenguaje natural (**NLP**). Contiene artículos de grupos de noticias organizados en **20 categorías**.
---

## **1. Características del Dataset**
- **Cantidad de documentos:** Aproximadamente **18,000 artículos**.
- **Número de categorías:** **20 grupos de noticias** temáticos.
- **Formato:** Textos sin procesar, incluyendo metadatos como encabezados, firmas y citas.

---

```python
# Importar librerías necesarias
import pandas as pd
import numpy as np
import re
import string
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
import nltk
# Descargar recursos de nltk si no están disponibles
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
newsgroups = fetch_20newsgroups(subset='train', categories=['alt.atheism', 'sci.space', 'comp.graphics'])
```

```python
print(f"Dataset con {len(newsgroups.data)} documentos y {len(newsgroups.target_names)} categorías.")
print(newsgroups.data[200])
```

```python
def preprocess_text(text):

  text = text.lower()
  text = re.sub(r'[^a-z\s]', ' ', text)
  #print(f"Esto es el código sin números y caracteres especiales: {text}")

  words = word_tokenize(text)
  #print(f"Esto es el código tokenizado: {words}")

  # Stopwords

  stop_words = set(stopwords.words('english'))

  words = [word for word in words if word not in stop_words]

  #print(f"Esto es el código sin stopwords: {words}")

  # Stemming

  stemmer = PorterStemmer()
  words_stemmed = [stemmer.stem(word) for word in words]

  #print(f"Esto es el código con stemming: {words_stemmed}")

  # Lemmatization

  lemmatizer = WordNetLemmatizer()

  words_lemmatized = [lemmatizer.lemmatize(word) for word in words]
  #print(f"Esto es el código con lemmatization: {words_lemmatized}")

  text_final = ' '.join(words_lemmatized)
  #text = re.sub(r'  ', ' ', text)
  #print(f"Esto es el text final: {text_final}")
  return text_final
```

```python
preprocess_text_data = [preprocess_text(text) for text in newsgroups.data]
```

```python
df = pd.DataFrame({'text': preprocess_text_data, 'target': newsgroups.target})
```

```python
vectorizers_bow = CountVectorizer(max_features=20, lowercase=False)
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['target'], test_size=0.2, random_state=21)
```

```python
vectorizers_bow.vocabulary_
```

```python
X_train = vectorizers_bow.fit_transform(X_train)
```

```python
X_train[0]
```

```python
X_test = vectorizers_bow.transform(X_test)
```

```python
from sklearn.naive_bayes import BernoulliNB
```

```python
model = BernoulliNB()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(classification_report(y_test, y_pred, target_names=['alt.atheism', 'sci.space', 'comp.graphics']))
```

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
import pandas as pd
import torch
```

```python
newsgroups = fetch_20newsgroups(subset='train', categories=['alt.atheism', 'sci.space', 'comp.graphics'])
texts = newsgroups.data[:100]
targets = newsgroups.target[:100]
```

```python
label_encoder = LabelEncoder()
labels = label_encoder.fit_transform(targets)
```

```python
X_train, X_test, y_train, y_test = train_test_split(texts, labels, test_size=0.2, random_state=21)
```

```python
train_df = pd.DataFrame({"text": X_train, "labels": y_train})
```

```python
test_df = pd.DataFrame({"text": X_test, "labels": y_test})
```

```python
model_type = 'cardiffnlp/twitter-roberta-base-sentiment-latest'
model = ClassificationModel("roberta", model_type, use_cuda=torch.cuda.is_available(), num_labels=len(set(labels)))
```

```python
model.train_model(train_df)
```

```python
test_results, model_outputs, wrong_predictions = model.eval_model(test_df)
print(f"Resultados: {test_results}")
```
